# 05 – Do the other inputs behave sensibly?
**Test:** for each remaining field, change only that field for all 600 applicants.

| Field | Expected |
|---|---|
| credit score | higher → fewer rejections |
| loan amount | bigger → more rejections |
| tenure | longer (smaller EMI) → fewer rejections |
| employment years | more → fewer rejections |
| age | little or no effect |

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
from common import load_data, sweep

df = load_data()
tests = {
    "credit_score": [300, 500, 600, 650, 700, 750, 800, 850, 900],
    "loan_amount": [50000, 100000, 200000, 500000, 1000000, 2000000, 4000000],
    "tenure_months": [12, 18, 24, 36, 48, 60],
    "employment_years": [0, 1, 2, 4, 6, 8, 10, 15, 20],
    "age": [21, 22, 23, 24, 25, 30, 40, 50, 62],
}
```

```python
for column, values in tests.items():
    table = sweep(df, column, values, probe_file=f"05_{column}_sweep.csv")
    print(column)
    display(table)
```

**Conclusion:**
- Credit score, loan amount, tenure and employment all move in the sensible direction.
- Tenure is flat between 36 and 48 months. Notebook 07 explains why.
- **Age is not flat:** 21–24-year-olds are rejected more (70% at 21 vs 57% from 25 on). Notebook 07
  digs into this.
